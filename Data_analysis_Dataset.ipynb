```python
import matplotlib.pyplot as plt

category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind="bar")

plt.title("Category-wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
- Electronics has the highest total sales among all categories.
- Furniture has the second Highest sales.
- Clothing has the lowest sales.
```

```python
df["Month"] = df["Order_Date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
- Octomber has the highest sales of the year.
- July has the lowest sales of the year.
- Sales fiuctuate throughout the year rather than increasing steadily.
- Sales increase significantly from September to Octomber.
- After Octomber, Sales drop in November and increase again slightly in December.
- April and Augast also show relatively strong sales.
- Overall, the graph shows that sales are seasonal/variable, with October being the Strongest month.
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Region-wise Sales")
plt.ylabel("")
plt.show()
```

```python
- South is the highest sales in region.
- North is the lowest sales in region.
```

```python
state_profit = df.groupby("State")["Profit"].sum().sort_values(ascending=False)

state_profit.plot(kind="bar")

plt.title("State-wise Profit")
plt.xlabel("State")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.show()
```

```python
- Maharashtra has the highest total profit.
- Gujarat has the second_highest profit.
- West Bengal has the lowest lowest profit among the states shown.
- Odisha, Tamil Nadu, Telangana, Karnataka, and Kerla are also strong contributors.
- 
```

```python
plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
- There is a Positive relationship between Sales and profit.
- When Sales increase, profit also increase.
```

```python
df.boxplot(
    column="Profit",
    by="Category"
)

plt.title("Category-wise Profit Distribution")
plt.suptitle("")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
- Electronics has the highest median profit among the three categories.
- Furniture has the second-highest median profit.
- Clothing has the lowest median profit.
- Electronics has a much wider profit distribution, meaning profit varies considerably between different Electronics orders.
- There are many outliers in Electronics and Furniture. These are orders with unusually high profit.
- Clothing has a relatively small profit range, although it also has some outliers.
```

```python
import seaborn as sns

numeric_cols = [
    "Quantity",
    "Unit_Price",
    "Discount",
    "Sales",
    "Profit"
]

corr = df[numeric_cols].corr()

sns.heatmap(corr, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

```python
- very strong positive relationship between Sales and Profit.
```
